# 1. IMPORT LIBRARIES

```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf

from tensorflow.keras import Sequential
from tensorflow.keras.layers import Input, Flatten, Dense, Dropout
from PIL import Image, ImageOps
from google.colab import files
import io

print("TensorFlow version:", tf.__version__)
```

 **2. LOAD MNIST DATASET**


```python
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.mnist.load_data()

print("Training images:", x_train.shape)
print("Training labels:", y_train.shape)
print("Testing images:", x_test.shape)
print("Testing labels:", y_test.shape)
```

 3. EXPLORE DATASET

```python
print("Number of classes:", len(np.unique(y_train)))
print("Classes:", np.unique(y_train))
print("Image size:", x_train[0].shape)
```

4. DISPLAY SAMPLE IMAGES

```python
plt.figure(figsize=(10, 5))

for i in range(10):
    plt.subplot(2, 5, i + 1)
    plt.imshow(x_train[i], cmap="gray")
    plt.title("Label: " + str(y_train[i]))
    plt.axis("off")

plt.tight_layout()
plt.show()
```

 5. NORMALIZE DATA

```python
x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0

print("Minimum pixel value:", x_train.min())
print("Maximum pixel value:", x_train.max())
```

 6. BUILD NEURAL NETWORK

```python
model = Sequential([
    Input(shape=(28, 28)),
    Flatten(),
    Dense(128, activation="relu"),
    Dropout(0.2),
    Dense(10, activation="softmax")
])
```

7. COMPILE MODEL

```python
model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

print("Model compiled successfully.")
```

8. TRAIN MODEL

```python
history = model.fit(
    x_train,
    y_train,
    epochs=10,
    batch_size=32,
    validation_split=0.1
)
```

9. EVALUATE MODEL

```python
test_loss, test_accuracy = model.evaluate(
    x_test,
    y_test,
    verbose=0
)

print("Test Loss:", test_loss)
print("Test Accuracy:", test_accuracy)
print("Test Accuracy Percentage:", test_accuracy * 100)
```

10. PLOT TRAINING/VALIDATION ACCURACY

```python
plt.figure(figsize=(8, 5))

plt.plot(
    history.history["accuracy"],
    label="Training Accuracy"
)

plt.plot(
    history.history["val_accuracy"],
    label="Validation Accuracy"
)

plt.title("Training and Validation Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()
plt.grid()
plt.show()
```

11. PLOT TRAINING/VALIDATION LOSS

```python
plt.figure(figsize=(8, 5))

plt.plot(
    history.history["loss"],
    label="Training Loss"
)

plt.plot(
    history.history["val_loss"],
    label="Validation Loss"
)

plt.title("Training and Validation Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.grid()
plt.show()
```

12. PREDICT TEST IMAGES

```python
predictions = model.predict(x_test, verbose=0)

predicted_labels = np.argmax(predictions, axis=1)

print("Actual labels:   ", y_test[:10])
print("Predicted labels:", predicted_labels[:10])
```

DISPLAY ACTUAL VS PREDICTED LABELS

```python
plt.figure(figsize=(10, 5))

for i in range(10):
    plt.subplot(2, 5, i + 1)

    plt.imshow(x_test[i], cmap="gray")

    plt.title(
        "Actual: " + str(y_test[i]) +
        "\nPredicted: " + str(predicted_labels[i])
    )

    plt.axis("off")

plt.tight_layout()
plt.show()
```

PREPROCESS HANDWRITTEN IMAGE

```python
def preprocess_image(file_data):

    img = Image.open(
        io.BytesIO(file_data)
    ).convert("L")

    # Invert image
    img = ImageOps.invert(img)

    # Crop unnecessary background
    bbox = img.getbbox()

    if bbox:
        img = img.crop(bbox)

    # Resize while maintaining proportions
    img.thumbnail((20, 20))

    # Create 28x28 black image
    canvas = Image.new("L", (28, 28), 0)

    # Center digit
    x = (28 - img.width) // 2
    y = (28 - img.height) // 2

    canvas.paste(img, (x, y))

    # Convert to NumPy array
    image_array = np.array(canvas)

    # Normalize
    image_array = image_array.astype("float32") / 255.0

    return image_array
```

```python
from google.colab import files
uploaded = files.upload()
```

PREDICT 5 HANDWRITTEN IMAGES

```python
results = []

for filename, file_data in uploaded.items():

    image = preprocess_image(file_data)

    prediction = model.predict(
        image.reshape(1, 28, 28),
        verbose=0
    )

    predicted_digit = np.argmax(prediction)

    # Actual digit is taken from filename
    actual_digit = int(filename.split(".")[0])

    results.append(
        (filename, actual_digit, predicted_digit)
    )
```

17. PRINT RESULTS

```python
print("Handwritten Image Results")
print("-------------------------")

for filename, actual, predicted in results:

    result = "Correct" if actual == predicted else "Wrong"

    print(
        "Image:", filename,
        "| Actual:", actual,
        "| Predicted:", predicted,
        "|", result
    )
```

18. DISPLAY 5 HANDWRITTEN IMAGES

```python
plt.figure(figsize=(12, 4))

for i, (filename, actual, predicted) in enumerate(results):

    image = preprocess_image(
        uploaded[filename]
    )

    plt.subplot(1, len(results), i + 1)

    plt.imshow(image, cmap="gray")

    plt.title(
        f"Actual: {actual}\nPredicted: {predicted}"
    )

    plt.axis("off")

plt.tight_layout()
plt.show()
```

19. EXPERIMENT CHANGE NUMBER OF NEURONS

```python
experiment_model = Sequential([
    Input(shape=(28, 28)),
    Flatten(),
    Dense(64, activation="relu"),
    Dropout(0.2),
    Dense(10, activation="softmax")
])

experiment_model.summary()
```

20. COMPILE EXPERIMENT MODEL

```python
experiment_model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)
```

21. TRAIN EXPERIMENT MODEL

```python
experiment_history = experiment_model.fit(
    x_train,
    y_train,
    epochs=10,
    batch_size=32,
    validation_split=0.1
)
```

22. EVALUATE EXPERIMENT MODEL

```python
experiment_loss, experiment_accuracy = (
    experiment_model.evaluate(
        x_test,
        y_test,
        verbose=0
    )
)

print(
    "Experimental Test Loss:",
    experiment_loss
)

print(
    "Experimental Test Accuracy:",
    experiment_accuracy
)

print(
    "Experimental Accuracy Percentage:",
    experiment_accuracy * 100
)
```

23. COMPARE BOTH MODELS

```python
print("\nMODEL COMPARISON")
print("-------------------------")

print(
    "Original Model - 128 neurons:",
    round(test_accuracy * 100, 2),
    "%"
)

print(
    "Experimental Model - 64 neurons:",
    round(experiment_accuracy * 100, 2),
    "%"
)

difference = (
    experiment_accuracy - test_accuracy
) * 100

print(
    "Difference:",
    round(difference, 2),
    "percentage points"
)
```

24. PLOT EXPERIMENT COMPARISON

```python
plt.figure(figsize=(8, 5))

plt.plot(
    history.history["val_accuracy"],
    label="Original - 128 Neurons"
)

plt.plot(
    experiment_history.history["val_accuracy"],
    label="Experiment - 64 Neurons"
)

plt.title("Validation Accuracy Comparison")
plt.xlabel("Epoch")
plt.ylabel("Validation Accuracy")
plt.legend()
plt.grid()
plt.show()
```

25. FINAL RESULT TABLE

```python
print("\nFINAL RESULTS")
print("-------------------------")
print(
    f"Original Model (128 neurons): "
    f"{test_accuracy * 100:.2f}%"
)
print(
    f"Experimental Model (64 neurons): "
    f"{experiment_accuracy * 100:.2f}%"
)
print(
    f"Difference: {difference:.2f} percentage points"
)
```
